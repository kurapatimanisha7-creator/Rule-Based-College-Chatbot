```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../dataset/intents.csv")

df.head()
```

```python
print("Total Questions:", len(df))
```

```python
print("Total Intents:", df["intent"].nunique())
```

```python
print("Defined Intents:")

for intent in df["intent"].unique():
    print("-", intent)
```

```python
intent_counts = df["intent"].value_counts()

print(intent_counts)
```

```python
print("Minimum examples:", intent_counts.min())
print("Maximum examples:", intent_counts.max())
```

```python
intent_counts.plot(kind="bar")

plt.xlabel("Intent")
plt.ylabel("Number of Questions")
plt.title("Intent Distribution")

plt.tight_layout()
plt.show()
```
